## Decide your cleaning strategy / Zdecyduj o strategii czyszczenia

For the two predictive paths (clinic wait times, loan default), the train/test split happens here, before anything is imputed — `impute_missing` only ever looks at `train_df`. For the segmentation path (retail stores), there's no held-out set to protect in this sense, so `clean_dataset` fills the whole dataset at once, exactly as before.

```python
from task import clean_dataset, impute_missing, split_dataset

if DATASET_NAME == "clinic_wait_times":
    train_df, test_df = split_dataset(df)
    features = ["num_patients_ahead", "staff_on_duty", "hour_of_day", "patient_age"]
    train_df, test_df = impute_missing(train_df, test_df, features)
    missing_value_counts(train_df)
elif DATASET_NAME == "lendwell_loan_default":
    train_df, test_df = split_dataset(df, stratify_column="defaulted")
    features = [
        "loan_amount",
        "annual_income",
        "credit_score",
        "debt_to_income_ratio",
        "employment_years",
        "previous_defaults",
    ]
    train_df, test_df = impute_missing(train_df, test_df, features)
    missing_value_counts(train_df)
elif DATASET_NAME == "retail_store_segments":
    cleaned = clean_dataset(df)
    missing_value_counts(cleaned)
else:
    raise ValueError(f"Unknown DATASET_NAME: {DATASET_NAME!r}")
```
