## How do the numeric features relate to each other?

`numeric_correlations` picks out only the numeric columns automatically and computes their pairwise correlation. For the two predictive paths, `df` above is `train_df` restricted to the suggested feature columns plus the target — the correlation matrix below never sees a row from `test_df`. For the segmentation path, `df` is the whole cleaned dataset, since there's no split to protect.

```python
from task import numeric_correlations

corr = numeric_correlations(df)
corr
```

## Which relationships stand out?

Look for the strongest correlations in the table above — both the highest positive and the lowest (most negative) values. (Every feature correlates perfectly with itself, so the diagonal's 1.0 values collapse to a single entry below — that's expected, not a bug.)

```python
corr.unstack().sort_values(ascending=False).drop_duplicates()
```
