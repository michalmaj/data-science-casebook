## Do the two metrics agree?

Inertia always goes down as k grows — that's expected. The question is whether it drops sharply at some particular k (an "elbow") or just fades out smoothly. Compare that to which k has the highest silhouette score.

```python
metrics.sort_values("silhouette", ascending=False)
```

## Is silhouette's runner-up even meaningful?

```python
metrics.set_index("k")["silhouette"].diff()
```
