## A different question: does KMeans's own randomness matter?

`subsample_stability` only varies which subscribers are sampled — KMeans itself always runs with the same `random_state`. `initialization_stability` asks a different question: on the *same* full data, does KMeans's random starting point change the answer?

```python
from task import initialization_stability

initialization_stability(df, k=2)
```

```python
for k in [2, 3, 4, 5]:
    result = initialization_stability(df, k=k)
    print(f"k={k}: min ARI across random_states = {result['adjusted_rand_index'].min():.6f}")
```
